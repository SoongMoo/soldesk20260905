```python
def calc(x1, x2, opt):
    if opt == "add":
        result = x1 + x2
    elif opt == "sub":
        result = x1 - x2
    elif opt == "mul":
        result = x1 * x2
    else:
        result = x1 / x2
    return result
```

```python
y = calc(3, 4, "add")
print(y)
y = calc(3, 4, opt = "add")
print(y)
# 매개변수의 이름을 사용하면 매개변수의 순서와 상관없이 값을 전달 할 수 있다.
y = calc(opt = "add", x1 = 3, x2 = 4)
print(y)

# y = calc(30, opt="add",  40)
# print(y)
y = calc(30, opt="add",  x2 = 40)
print(y)
```

```python
def create_profile(** args):
    print(args.items())
    for key in args.keys():
        print(key , args[key])
    for key, value in args.items():
        print(key, value)
    for key, value in zip(args.keys(), args.values()):
        print(key, value)
```

```python
b = 100 # 전역변수
def test2():
    global b # 전역변수 b사용
    c1 = 200 # 지역변수
    b = 200 # 전역변수
    print("지역 변수 : " , b)
    print("지역 변수 : ", c1)
test2()
print(b)
```

```python
f = open("test.txt", 'w', encoding="utf-8") # overwrite
for num in range(6, 11):
    f.write(f"{num}번째줄입니다. \n")
f.close()
```

```python
f = open("test.txt", 'a', encoding="utf-8") #append
for num in range(11, 16):
    f.write(f"{num}번째줄입니다. \n")
f.close()
```

```python
f = open("test.txt",'r', encoding="utf-8") 
line = f.readline()
print(line)
line = f.readline()
print(line)
line = f.readline()
print(line)
line = f.readline()
print(line)
print(line)
f.close()
```

```python
f = open("test.txt",'r', encoding="utf-8") 
while True:
    line = f.readline() # "" : False
    if not line: # not False == True
        break
    print(line)
f.close()
```

```python
f = open("test.txt",'r', encoding="utf-8") 
lines = f.readlines()
f.close()
print(lines)
print("".join(lines))
```

```python
f = open("test.txt",'r', encoding="utf-8") 
data = f.read()
f.close()
print(data)
```

```python
with open("test.txt",'w', encoding="utf-8") as f:
    for num in range(1, 6):
        f.write(f"{num}번째 줄입니다.\n")
```

```python
with open("test.txt",'a', encoding="utf-8") as f:
    for num in range(6, 11):
        f.write(f"{num}번째 줄입니다.\n")
```

```python
with open("test.txt",'r', encoding="utf-8") as f:
    while True:
        line = f.readline()
        if not line:
            break
        print(line)
```

```python
with open("test.txt",'r', encoding="utf-8") as f:
    lines = f.readlines()
print(lines)
print("".join(lines))
```
