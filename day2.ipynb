```python
# lotto
import random
lotto = []
i = 1
while i < 46:
    lotto.append(i)
    i += 1
print(lotto)
size = len(lotto) # 45
num = 1
while num < 7:
    size = size - 1
    idx = random.randint(0, size) # 0 ~ 44
    result = lotto.pop(idx)
    #print(idx , result)  
    print(result , end =", ")    
    num += 1
```
