```python
# 딕셔너리를 이용한 시리즈
import pandas as pd
sdata = pd.Series({"Ohio":35000, "Texas":71000, "Oregon":16000, "Utah":50000})
print(sdata)
print(sdata['Ohio'])
print(sdata.iloc[0])
dic = {"Ohio":35000, "Texas":71000, "Oregon":16000, "Utah":50000}
sdata = pd.Series(dic)
# index의 순서를 변경 하자: index에 있는 것만 시리즈에 저장
sdata = pd.Series(dic , index=["Oregon","Ohio","Utah"])
print(sdata)
sdata = pd.Series(dic , index=["California","Oregon","Ohio","Utah"])
print(sdata)
# 시리즈와 index에 이름 주기
sdata.name="이숭무"
print(sdata)
sdata.index.name="이상범"
print(sdata)
#  시리즈가 만들어진후 index 변경
sdata.index=['Oregon','Utah','Ohio','Texas']
print(sdata)
```

```python
# 누락된 데이터(결측치 : NaN) 처리
data = pd.Series([1, np.nan, 3.5, np.nan, 7])
print(data)
print(data.isna())
# 누락값 제외하고 출력
print(data.dropna())
print(data[data.notna()])
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])
print(data)
# 누락값 제외하고 출력
print(data.dropna(how="all")) # 모두  NaN인 행만 제거
print(data.dropna(how="all", axis="index"))
print(data.dropna(how="all", axis="columns"))
```

```python
# 적어도 하나가 NaN이면 제거
print(data.dropna())
print(data.dropna(how="any"))
print(data.dropna(how="any", axis="index"))
print(data.dropna(how="any", axis="columns"))
```

```python
# 결측치 값을 채우기
print(data)
print(data.mean()) # 각 열에 대한 평균
print(data.fillna(data.mean()))
print(data.fillna(method="ffill"))
print(data.fillna(method="bfill"))
# 연속된 NaN 값이 있을 때, 최대 2개까지만 앞의 값으로 채워줍니다
print(data.fillna(method="bfill", limit=2))
```

```python
# group by
df = pd.read_csv("https://raw.githubusercontent.com/SoongMoo/data-analysis/refs/heads/main/examples/gapminder.tsv"
                , sep="\t")
df
```

```python
df.groupby('year')['lifeExp'].max()
df.groupby('year').lifeExp.mean()
df.groupby('year').gdpPercap.max()
df.groupby('year')['gdpPercap'].max()
```

```python
df.groupby(['continent','country'])['lifeExp'].max()
df.groupby(['continent','country'])['lifeExp'].min()
df.groupby(['continent','country'])['lifeExp'].mean()
# 각 년도별 대륙의 평균수명의 최소값과 gdp의 최소값을 출력
df.groupby(['year','continent'])[['lifeExp','gdpPercap']].min()
```
