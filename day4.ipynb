```python
# 상속 : 이미 만들어져 있는 것을 가져와서 재사용
class Calculator: # 부모 클래스
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        self.result = self.first + self.second
    def div(self):
        self.result = self.first / self.second
```

```python
class FourCalc(Calculator): # 자식클래스
    # def __init__(self, first, second):
    #     self.first = first
    #     self.second = second
    # def add():
    #     self.result = self.first + self.second
    # def div():
    #     self.result = self.first / self.second
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
```

```python
four = FourCalc(100, 200)
four.add()
print(four.result)
four.sub()
print(four.result)
```

```python
four = FourCalc(0, 200) # 0 / 200 = 0
four.div()
print(four.result)
```

```python
four = FourCalc(200, 0) # 200 / 0 = 오류
four.div()
print(four.result)
```

```python
class FourCalc(Calculator): # 자식클래스
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
    def div(self): # overriding: 부뮤클래스메서드를 자식클래스에서 다시 정의하는 것
        if self.second == 0:
            self.result = 0
        else:
            self.result = self.first / self.second
```

```python
four = FourCalc(200, 0) # 200 / 0 = 오류
four.div()
print(four.result)
```

```python
# Boolean Indexing
names = np.array(["Bob", "Joe", "Will", "Bob", "Will", "Joe", "Joe"])
data = np.array([[4, 7], [0, 2], [-5, 6], [0, 0], [1, 2],[-12, -4], [3, 4]])
print(names)
print(data)
print(names == "Bob")
#[ True False False  True False False False]:Boolean Indexing
print(data[[ True, False, False,  True, False, False, False]])
print(data[names == "Bob"])

print(names != "Bob")
print(data[names != "Bob"])
print(data[~(names == "Bob")])

mask = (names == "Bob") | (names == "Will")
print(mask)
print(data[mask])
```

```python
import pandas as pd
rainfall = pd.read_csv("https://raw.githubusercontent.com/SoongMoo/data-analysis/refs/heads/main/data/Seattle2014.csv")
rainfall # PRCP:강수량
# 열 데이터를 배열로 받아옴
inches = np.array(rainfall['PRCP'] / 254)
inches
# 0.5인치이상 온 날은 몇일입니까?
inches >= 0.5 # False= 0, True = 1
print(np.sum(inches >= 0.5))
# 강수량이 0.5에서 0.8 사이인 날은 몇일이나 되나요?
print(np.sum((inches >= 0.5) & (inches <= 0.8)))
#강수량이 0.5보다 작고 0.8보다는 많이 온날은?
print(np.sum(~((inches >= 0.5) & (inches <= 0.8))))
#비가 안온 날
print(np.sum(inches == 0))
#비가 온 날
print(np.sum(inches != 0))
```

```python
# 일반 반복문
xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])
cond = np.array([True, False, True, True, False])
result = []
for x, y, c in zip(xarr, yarr, cond):
    if c:
        result.append(float(x))
    else:
        result.append(float(y))
# 리스트 컨프리핸션으로 변경
result = [ float(x if c else y)
           for x, y, c in zip(xarr, yarr, cond)]
print(result) 
```
