```python
# 상속 : 이미 만들어져 있는 것을 가져와서 재사용
class Calculator: # 부모 클래스
    def __init__(self, first, second):
        self.first = first
        self.second = second
    def add(self):
        self.result = self.first + self.second
    def div(self):
        self.result = self.first / self.second
```

```python
class FourCalc(Calculator): # 자식클래스
    # def __init__(self, first, second):
    #     self.first = first
    #     self.second = second
    # def add():
    #     self.result = self.first + self.second
    # def div():
    #     self.result = self.first / self.second
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
```

```python
four = FourCalc(100, 200)
four.add()
print(four.result)
four.sub()
print(four.result)
```

```python
four = FourCalc(0, 200) # 0 / 200 = 0
four.div()
print(four.result)
```

```python
four = FourCalc(200, 0) # 200 / 0 = 오류
four.div()
print(four.result)
```

```python
class FourCalc(Calculator): # 자식클래스
    def sub(self):
        self.result = self.first - self.second
    def mul(self):
        self.result = self.first * self.second
    def div(self): # overriding: 부뮤클래스메서드를 자식클래스에서 다시 정의하는 것
        if self.second == 0:
            self.result = 0
        else:
            self.result = self.first / self.second
```

```python
four = FourCalc(200, 0) # 200 / 0 = 오류
four.div()
print(four.result)
```
